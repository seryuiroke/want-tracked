# discussion and economic insights

the previous notebook explored how consumer attention changed across several trends over time.

this notebook interprets those findings through concepts from economics and consumer behavior. rather than describing what happened, the goal is to explore why different trends experience different levels of attention and what these patterns may suggest about consumer demand in the attention economy.

## what is the attention economy?

in digital markets, attention has become a limited resource. consumers are exposed to thousands of products, brands, and experiences every day, while companies compete to capture only a small amount of that attention.

online activity such as pageviews, searches, and social media engagement cannot directly measure purchases, but they can provide signals about changing consumer interest. understanding these patterns helps explain how demand develops before purchasing decisions are made.

## why do some trends receive more attention?

the exploratory analysis showed that different trends follow different paths over time.

some, such as matcha, maintained consistently high attention throughout the study period. others, including jorts, experienced much faster percentage growth despite attracting a much smaller audience overall.

these differences suggest that consumer demand develops in multiple ways. some products become established parts of everyday consumption, while others spread rapidly through short-term popularity or changing cultural preferences.

## consumer demand and social proof

consumer demand is influenced by more than price or product quality. people are also affected by social proof, where the popularity of a product encourages additional interest from others.

digital platforms allow trends to spread quickly because consumers constantly observe what other people are purchasing, wearing, or discussing online. as more attention is directed toward a product, additional consumers may become interested, creating a cycle that reinforces demand.

## scarcity and perceived value

scarcity can increase demand by making a product or experience feel more exclusive. limited releases, sold-out products, and short availability windows may create urgency because consumers believe they could miss the opportunity to participate in a trend.

in digital markets, scarcity can be real or perceived. a product may be physically limited, or it may simply appear difficult to obtain because of viral demand, long lines, resale activity, or repeated posts showing that it is unavailable. this can increase perceived value even when the product itself has not changed.

## network effects and cultural diffusion

trends often become more valuable as more people participate in them. this is related to network effects, where the usefulness or appeal of something increases as its audience grows.

social platforms accelerate this process by allowing styles, products, and experiences to move quickly across communities and locations. a trend associated with tokyo or new york can spread beyond those places as users reproduce the look, share recommendations, and connect the trend to a larger cultural identity.

this project therefore treats japan and new york as cultural reference points rather than direct causes of demand.

## interpreting the current trend patterns

the results suggest that popularity, growth, and longevity should be considered together.

matcha received the highest average attention, but its negative growth indicates that high overall interest does not always mean that a trend is still expanding. jorts showed the greatest percentage growth, but it began with a much smaller level of attention. uniqlo and onitsuka tiger showed positive growth with larger audiences, which may reflect more gradual increases in brand awareness.

these differences show why one metric cannot fully represent consumer demand. a fast-growing trend may still be relatively small, while an established trend may remain economically relevant even after its attention begins to decline.

```python
import pandas as pd
import matplotlib.pyplot as plt

trend_profiles = pd.read_csv("../data/trend_profiles.csv")

display_names = {
    "digital_cameras": "Digital Cameras",
    "jorts": "Jorts",
    "matcha": "Matcha",
    "onitsuka_tiger": "Onitsuka Tiger",
    "uniqlo": "Uniqlo"
}

trend_profiles["display_name"] = (
    trend_profiles["trend_id"].map(display_names)
)

plt.figure(figsize=(9, 6))

plt.scatter(
    trend_profiles["average_attention"],
    trend_profiles["percent_growth"],
    s=90
)

for _, row in trend_profiles.iterrows():
    plt.annotate(
        row["display_name"],
        (
            row["average_attention"],
            row["percent_growth"]
        ),
        xytext=(6, 6),
        textcoords="offset points"
    )

plt.axhline(0, linewidth=1)
plt.title("Popularity and Growth in Consumer Attention")
plt.xlabel("Average Monthly Wikipedia Pageviews")
plt.ylabel("Percent Growth (%)")

plt.tight_layout()
plt.show()
```

### interpretation

the relationship between popularity and growth shows that consumer attention cannot be represented by one measure alone.

matcha attracted the largest average audience but experienced declining attention by the end of the study period. jorts showed the strongest percentage growth but remained much smaller in overall attention. uniqlo and onitsuka tiger combined positive growth with larger audiences, suggesting more gradual increases in public interest.

this distinction is economically important because a large established market and a rapidly growing emerging market can both be valuable, but they represent different forms of consumer demand.

## limitations

this project uses wikipedia pageviews as an indicator of public attention rather than direct consumer purchases. while pageviews provide valuable insight into changing interest over time, they cannot measure actual sales or explain why attention changed.

external events such as marketing campaigns, celebrity influence, seasonal demand, or news coverage may also affect attention. because of this, the results should be interpreted as indicators of consumer interest rather than proof of purchasing behavior.
