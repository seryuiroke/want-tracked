## planned trend measurements

each selected trend will eventually receive a data-driven profile based on its attention pattern over time.

the initial measurements will include:

- **growth:** how much attention increases during a trend's rise
- **peak:** the period with the highest observed attention
- **longevity:** how long elevated attention persists
- **decline:** how quickly attention falls after its peak
- **volatility:** how dramatically attention changes over time
- **momentum:** whether recent attention is increasing, stable, or decreasing

these measurements will be used to compare trends without assuming that search or online attention is equivalent to purchasing behavior.

```python
import requests
```

```python
def get_wikipedia_pageviews(article, start="20210101", end="20260701"):
    article = article.replace(" ", "_")

    url = (
        "https://wikimedia.org/api/rest_v1/metrics/pageviews/per-article/"
        f"en.wikipedia.org/all-access/user/{article}/monthly/{start}/{end}"
    )

    headers = {
        "User-Agent": "want-tracked-data-project/1.0"
    }

    response = requests.get(url, headers=headers)
    response.raise_for_status()

    data = response.json()["items"]

    return pd.DataFrame({
        "date": pd.to_datetime(
            [item["timestamp"][:6] for item in data],
            format="%Y%m"
        ),
        "views": [item["views"] for item in data],
        "article": article.replace("_", " ")
    })
```

```python
onitsuka_views = get_wikipedia_pageviews("Onitsuka Tiger")

onitsuka_views.tail()
```

```python
onitsuka_views = onitsuka_views[
    onitsuka_views["date"] < "2026-07-01"
]

onitsuka_views.tail()
```

```python
articles = {
    "onitsuka_tiger": "Onitsuka Tiger",
    "uniqlo": "Uniqlo",
    "matcha": "Matcha",
    "digital_cameras": "Digital camera",
    "jorts": "Jean shorts"
}

print("ready to collect", len(articles), "more trends")
```

```python
all_pageviews = []

for trend_id, article in articles.items():
    try:
        pageviews = get_wikipedia_pageviews(article)

        # remove incomplete july 2026 observations
        pageviews = pageviews[pageviews["date"] < "2026-07-01"]

        # identify each trend in the combined dataset
        pageviews["trend_id"] = trend_id

        all_pageviews.append(pageviews)

        print(f"✓ successfully collected {article}")

    except Exception as e:
        print(f"✗ unable to collect {article}")
        print(e)
```

## creating trend profiles

the monthly pageview dataset contains detailed observations for each trend over time. to better compare trends, the data will be summarized into a trend profile.

each profile captures key measures of consumer attention, including average attention, peak attention, growth over time, and volatility. these metrics make it possible to compare different types of trends and explore how online attention relates to consumer demand.

```python
trend_profiles = (
    pageviews_df
    .groupby("trend_id")
    .agg(
        average_attention=("views", "mean"),
        peak_attention=("views", "max"),
        lowest_attention=("views", "min"),
        total_attention=("views", "sum"),
        volatility=("views", "std")
    )
    .round(2)
)

trend_profiles
```

### measuring trend growth

average attention and peak attention describe how popular a trend is, but they do not show whether interest is increasing or decreasing over time.

to measure changes in consumer attention, the first and last observations for each trend will be compared. this provides a simple measure of growth that can later be used to discuss changes in consumer demand.

```python
growth = (
    pageviews_df
    .groupby("trend_id")
    .agg(
        first_attention=("views", "first"),
        latest_attention=("views", "last")
    )
)

growth["percent_growth"] = (
    (
        growth["latest_attention"] - growth["first_attention"]
    ) / growth["first_attention"]
) * 100

growth = growth.round(2)

growth
```

```python
trend_profiles = trend_profiles.join(growth)

trend_profiles
```

```python
trend_profiles.to_csv(
    "../data/trend_profiles.csv",
    index=True
)

print("trend profiles saved!")
```
